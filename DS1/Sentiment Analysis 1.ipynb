IMPORT LIBRARIES

```python
# This line of code will import the pandas library into the jupyter notebook. 
# Pandas is a python package that we will use to analyse and manipulate our dataset.
# Using the 'as' syntax, we were able to assign the alias 'pd' to panda to reference its functions easier throughout the script.

import pandas as pd
```

LOAD DATASET

```python
# This line of code loads our dataset, which is a CSV file named 'twitter_training.csv,' into a DataFrame called 'df.' 
# According to W3Schools, a DataFrame is a 2D data structure which can be visualized with a table with rows and columns.
# The function pd.read_csv() reads our dataset and automatically determines the column headers and data types for the creation of the DataFrame df.

df = pd.read_csv('twitter_training.csv')
```

INITIAL UNDERSTANDING

```python
# The function head() displays the first five rows of the df DataFrame. 
# This is a usual way for us to quickly inspect the column headers and their content.
# Through this, we are able to get a quick understanding of the structure of the data we are working with without loading all of it into memory.

df.head()
```

```python
# The shape() function is used to return a tuple showing the dimensionality of the DataFrame df (pandas, 2024).
# The first value 74 682 is the number of records in our dataset (rows), while the second value 2 is the number of columns.
df.shape
```

```python
# The info() function from the pandas library prints more information about our df DataFrame (pandas, 2024). 
# It includes information like the numbers of rows (indexed from 0 to 61 691), the number of columns (2), the different column names ('label' and 'premise'), the number of non-null rows, the data type of each columm ('object'), and the total memory usage of the dataset in RAM.
# It is important to note that the number of non-null values in the premise column is larger than the total number of records. This means that there are some values in the premise column that are null and needs to be cleaned later.

df.info()
```

```python
# This line of code is used to list all column names in the DataFrame df.
# Throught this, we were able to confirm that expected columns like 'label' and 'premise' are present in the DataFrame.

df.columns
```

```python
# This prints all the unique values in the 'label' column, identifying all sentiment categories in the dataset. 
# For the purposes of this analysis, we have elected to delete all records with 'irrelevant' values because they do not contribute to our task.

print(df['label'].unique())
```

```python
print(df['label'].value_counts())
```

```python
df = df[df['label'].str.lower() != 'irrelevant']
```

```python
print(df['label'].value_counts())
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.shape
```

```python
df.info()
```

```python
df[df['premise'].isnull()]
```

```python
df = df.dropna(subset=['premise'])
```

```python
df.info()
```
