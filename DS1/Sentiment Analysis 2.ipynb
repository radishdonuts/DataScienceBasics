```python
# This line of code filters out rows where the label is 'irrelevant.'
# We used the function str.lower() to ignore case sensitivity in the labels.
# This is an important cleaning step so that our dataset is only focused on valid sentiment categories.
    
df = df[df['label'].str.lower() != 'irrelevant']
```

```python
# This line of code displays the records with null values in the premise column.
# The 'df['premise']' part specifies the relevant column, while the 'isnull() function specifies that only null values are to be displayed.

df[df['premise'].isnull()]
```

```python
# The function dropna() removes all rows from the df DataFrame where the value in the 'premise' column is 'NaN.'
# The subset parameter ensure that only the 'premise' columns is checked for any null values and the other column 'label' is not checked for null values.
# This line also assigns the results back to the df DataFrame to filter out and records with missing text data.

df = df.dropna(subset=['premise'])
```

```python
# Similar to its usage above, the reset_index() function resets the indeces in the df DataFrame.
# The parameters 'drop' and 'inplace' also makes sure to avoid adding the old index as a new column in the DataFrame and applies the changes to the DataFrame itself.
    
df.reset_index(drop=True, inplace=True)
```

```python
# This final line of code provides a short summary of the different features of our df DataFrame.
# There are now less entries because we cleaned our dataset from null values and 'irrelevant' records.
    
df.info()
```

DATA PREPROCESSING

```python
#This line of code converts all tweets under the columns 'premise' to lowercase.
#Lowercasing unifies data, reducing the need to process the text's tone from its casing (Vanga, 2024).

df['premise'] = df['premise'].str.lower()
```

```python
#This function "str.replace()" aims to remove punctuation and special characters through replacing "f"[{string.punctuation}]"" with a blank string ("").
#The parameter "regex=True" considers the "string.punctuation" as a regular expression for easier pattern matching/searching within strings (Yang, 2020).
#Removal of punctuation further reduces tone ambiguity from its characters (Vanga, 2024).

df['premise'] = df['premise'].str.replace(f"[{string.punctuation}]", "", regex=True)
```

```python
#The English set of stopwords from the NLTK library is stored in the variable 'stop_words' (Geeksforgeeks, 2024).
#This set will be used later to filter out the stopwords under the column 'premise'.

stop_words = set(stopwords.words('english'))
```

```python
#This code applies a function to all rows of the 'premise' column to remove stopwords from each tweet.
#For each row, the tweet is split into individual words, stopwords are filtered out, and remaining words are rejoined into a string (Geeksforgeeks, 2024a).
#The new cleaned string is stored in a new column 'cleaned'.

df['cleaned'] = df['premise'].apply(lambda x: ' '.join(word for word in word_tokenize(x) if word.lower() not in stop_words))
```

```python
#This code simply prints the data frame to observe and compare its preprocessed and cleaned data before splitting it.
#The 'cleaned' column in here is not present in the CSV file itself, but only in the dataframe.

print(df)
```

DATA SPLITTING

```python
#This code assigns the 'cleaned' column to variable 'X' to prepare for data splitting later.
#This column contains the cleaned and preprocessed tweets that have gone through lowercasing and punctuation and stopwords removal.

X = df['cleaned']
```

```python
#This code assigns the 'label' column to variable 'y' to prepare for data splitting later.
#This column contains the sentiments for each tweet, in which the model will learn and aim to predict later.

y = df['label']
```

```python
#The function train_test_split() splits the data (X and y) into train and test subsets, with 80% for train (train_size=0.8) and 20% for test (test_size=0.2).
#The random_state parameter is for data shuffling before the split and stratify helps split the data in proportion to the 'label' column (Scikit Learn, 2018).
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, train_size=0.8, random_state=3, stratify=y)
```

FEATURE EXTRACTION

```python
#This initializes the vectorizer by creating an instance.
#The instance will be used to fit and transform the cleaned data later.
vectorizer = TfidfVectorizer()
```

```python
#The 'fit_transform' method helps learn the vocabulary and the IDF from the corpus.
#This then return the TF-IDF matrix to the variable X (Geeksforgeeks, 2024).
#TF-IDF stands for Term Frequency - Inverse Document Frequency which helps in scoring a word.
#This reveals the importance of a word by contrasting the commonality and rarity of a word (Karabiber, 2024).

X_train_trans = vectorizer.fit_transform(X_train)
```

```python
#The 'transform' method helps transform our test data in its standardized form (Khanna, 2020). 
#This applies the vocabulary learned from the training data.

X_test_trans = vectorizer.transform(X_test)
```

TRAINING THE MODEL

```python
#Logistic Regression is used for our model training, which is commonly known for its simple and reliable binary classification tasks through relationship analysis of given variables (Geeksforgeeks, 2024b).
#The maximum number of iterations is set to 10,000 to find the most accurate sentiment (label) prediction from the provided tweets (premises) (Scikit Learn, 2025).

model = LogisticRegression(max_iter=10000)
```

```python
#The fit() method is the training part of the model using the training data.
#The model learns from this data to prepare itself for later predictions.

model.fit(X_train_trans, y_train)
```

MAKING PREDICTIONS

```python
#The predict() methods is the prediction part of the model using the test data.
#Resulting predictions is stored in y_pred for performance evaluation later.

y_pred = model.predict(X_test_trans)
```

```python
#This prints the accuracy score of the model based on its predictions (y_pred) compared with the original labels (y_test).
#The accuracy score of our model is 0.8259304703476482, which reveals an 82.59% accuracy for this dataset. 

print(accuracy_score(y_test, y_pred))
```

```python
#This classification report reveals that our model predicts best on 'negative' tweets with the highest f1-score of 0.85, 'positive' predictions are high as well, with 0.83.
#However, this leaves 0.79 for the neutral tweets, which also reveals it struggles to identify this sentiment and would need further improvements.

print(classification_report(y_test, y_pred))
```
