TECHNIQUE A: Automatic Thresholding (Otsu's Method)

TECHNIQUE B: Adaptive Thresholding:

```python
#This method uses adaptive thresholding, wherein the threshold value is calculated, giving different thresholds for each region (TutorialsPoint, 2025).
#The adaptive method used here is the 'ADAPTIVE_THRESH_MEAN_C' that gets the mean of the neighborhood area.
#Based on the other parameters, the block size for the neighborhood is 5, while the constant used for calculation is 9.
#Once again, the resulting image was done with a trial and error of the parameters
adapt_thresh = cv2.adaptiveThreshold (bltl, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 5, 9)

#This function specifies the size of the image to be displayed.
#This creates a figure and uses figsize to specify the width and height in inches.
plt.figure(figsize=(12, 6))
#This function specifies which stored image will be displayed, in this case, it is the Adaptive Thresholding Image.
#Cmap is still present in the 'imshow' parameter as this is a grayscale image, without this the image would appear yellowish rather than grayscale.
plt.imshow(adapt_thresh, cmap='gray')
#This function adds a title to the displayed image.
#In this title, it adds a description and context to its viewer.
plt.title('Adaptive Thresholding')
#Since pyplot mainly works with plot generation, normally axes will be seen.
#This function turns of the display of the axes, mainly to focus on the image displayed.
plt.axis('off')
#This is the last function to be executed after all the plotting commands.
#This basically shows the plotted image and the result of all our plotting commands done earlier.
plt.show()
```

1) When would you use adaptive thresholding instead of simple (Otsu's) thresholding?

Adaptive Thresholding takes into consideration the neighbor pixels, thus this is better for images with non-uniform lighting conditions and complex intensity distributions (Buhl, 2023). A simple thresholding is best suited for uniform images that creates clear bimodal histograms. The images used in this activity has varying light and color intensities, thus adaptive thresholding is better for this.

2) Explain the roles of blockSize and C in adaptiveThreshold.

Block Size in the parameters helps define the size of the neighborhood area, in the case above, it is 5. This creates a 5x5 window where the threshold is actually calculated. On the otherhand, the C parameter is a constant that is subtracted from the mean calculated (TutorialsPoint, 2025). This is meant for fine-tuning the threshold.

TECHNIQUE C: Edge Detection (Canny)

```python
#Canny Edge Detection is utilized here for noise reduction and edge thinning (GeeksforGeeks, 2016).
#The parameters of the function below uses 100 as the lower threshold and 300 for the upper threshold.
#Any pixels with gradient magnitude lower than 100 are not considered edges, anything in between 100 and 300 are weak edges, and anything above 300 are strong edges.
canny = cv2.Canny(bltl, 100, 300)

#This function specifies the size of the image to be displayed.
#This creates a figure and uses figsize to specify the width and height in inches.
plt.figure(figsize=(12, 6))
#This function specifies which stored image will be displayed, in this case, it is the Canny Edge Detection Image.
#Cmap is still present in the 'imshow' parameter as this is a grayscale image, without this the image would appear yellowish rather than grayscale.
plt.imshow(canny, cmap='gray')
#This function adds a title to the displayed image.
#In this title, it adds a description and context to its viewer.
plt.title('Edge Detection')
#Since pyplot mainly works with plot generation, normally axes will be seen.
#This function turns of the display of the axes, mainly to focus on the image displayed.
plt.axis('off')
#This is the last function to be executed after all the plotting commands.
#This basically shows the plotted image and the result of all our plotting commands done earlier.
plt.show()
```

1) Describe the multi-stage process of the Canny edge detector.
   
The Canny Edge Detection undergoes multiple processes of (1)noise reduction, (2) intensity gradient identification, (3) non-maximum supression, and (4) hysteresis thresholding (GeeksforGeeks, 2016). 

First, noise reduction is done by blurring the image through a Gaussian filter. Second process involves finding the intensity gradient of the image using a Sobel operator. This helps identify both the horizontal and vertical gradients to attain the magnitude of the edges and its direction. Third process is the non-maximum suppression that preserves the sharpest edges through thinning them down. This is done by comparing the pixel with its neighbors and its gradient direction, after which the decision whether it is suppressed or kept. The final process in the Canny Algorithm is the Double Thresholding to filter the weak, strong, and non-edges to reveal prominent details and edges, while keeping weak edges that contribute to the strong ones.

2) What is the significance of the two threshold values (threshold1 and threshold2)?

The significance of the two threshold values are for the final stage of the edge detection process. It has been mentioned above that weak, strong, and non-edges are considered. Thus the threshold1 (lower threshold) sets a filter or a boundary to which pixels are not considered edges (below threshold1) and weak edges (above threshold1 but below threshold2). And the threshold2 (upper threshold) sets a boundary to which pixels will be considered as weak edges (above threshold1 but below threshold2) and strong edges (above threshold2).


TECHNIQUE D: Color Filtering (HSV)

```python
#Color Filtering utilizes a given HSV value of a color to separate the regions of an image (GeeksforGeeks, 2019c).
#This line of code converts the rgb_img with the RGB color space to an HSV one.
hsv_img = cv2.cvtColor(rgb_img, cv2.COLOR_RGB2HSV)

#This block of code focuses mainly on the darker parts of the image (mostly clothes).
#Below, numpy is used to specify and store the HSV values of the colors in the image in an array.
#It is observable that this block of code has a lower and upper range, this captures the color along the given values.
#This HSV color range was actually a mix of trial and error along with the searching of the appropriate HSV values for the color desired to capture.
#Lots of color ranges were tried here along with red, orange, and even green, however only utilizing the two colors provided in this code has preserved most of the significant part of the picture and disregards the background.
lower_dark = np.array([0, 0, 0])
upper_dark = np.array([180, 100, 70])

#This block of code focuses mainly on the lighter parts of the image (mostly skin).
#Below, numpy is used to specify and store the HSV values of the colors in the image in an array.
#It is observable that this block of code has a lower and upper range, this captures the color along the given values.
#The color range for this actually captures majority of the image, even the background, as the skin of the 2 figures in the image is in the same range as the background.
#Furthermore, the flowers below and details in the mountains were also captured due to a similar color range.
lower_light = np.array([0, 20, 80])
upper_light = np.array([25, 150, 255])

#These two lines of code does the same thing: masking.
#Despite similar purposes, the first line of this block masks the darker parts of the image, which mostly captures the clothes of the two figures in the image.
#The second line masks the lighter parts of the image including the hair, the flowers, and the orange-y sky background.
#The function 'inRange' helps in thresholding the image based on color ranges provided (upper and lower), creating a mask that falls within the captured color regions.
mask_dark = cv2.inRange(hsv_img, lower_dark, upper_dark)
mask_light = cv2.inRange(hsv_img, lower_light, upper_light)

#This line of code combines the two masks together, the lighter mask and the darker mask.
#The 'bitwise_or' helps perform the bitwise OR operation the two masks together, joining even their differences.
combined_mask = cv2.bitwise_or(mask_light, mask_dark)

#This line of code performs the bitwse AND operation to the inputs, as both of the inputs are both rgb_img, it returns the same image.
#However, the mask parameters provides a mask to the original 'rgb_img', thus printing an RGB Image of the white spaces in the mask.
highlighted_img = cv2.bitwise_and(rgb_img, rgb_img, mask=combined_mask)
```

```python
#Erosion, on the other hand, is used for noise removal of white noises which detaches white spaces.
#As observed, it increases the black spaces of the image and thins the white edges or the object area.
#The erode function below also utilizes the set kernel in one iteration, but this time processes the dilated image from above.
img_erode = cv2.erode(combined_mask, kernel, iterations=1)
```

```python
#As observed, the dilate function has increased white spaces, or the object area, because of this this operation is also useful in joining broken parts or edges in an object (GeeksforGeeks, 2017a).
#The dilate function utilizes the kernel, in this case 3x3, and dilates the combined_mask (previously used in color filtering) for one iteration.
img_dilate = cv2.dilate(img_erode, kernel, iterations=1)
```

1) Explain the difference between erosion and dilation.

Erosion removes small white noises and even separates objects, while dilation increases the object area (white spaces) or connects objects. From observation, I would differentiate them as erosion adding more black spaces in the image and dilation adding more white spaces in the image. In a general perspective, these two help refine the image through noise removal and safer object ara expansion. 

2) How would applying these operations in sequence (e.g., erosion then dilation, or dilation then erosion) affect the image?

The sequences in applying the operations display different results as each operation serve different purposes. An opening operation is intended to remove small white noises with no worry to large object areas as they are preserved. This means that erosion shrinks object area to lessen white noises then dilation increases the area with less white noise back to its original size. A closing operation's goal is different, which initially uses dilation to fill in the black areas within object areas, then smoothens the contour of the dilated spaces once erosion takes place. The sequences matter but this depends on the image on what it needs to result to a more refined image.

3) What are "opening" and "closing" operations?

Opening operation is erosion then dilation, while closing operation is dilation then erosion. Each operation has a different goal, thus making the image have varying operation to execute depending on its need. The opening operation is mainly for white noise removal, while the closing operation closes holes or gaps in object areas.

```python
#This line of code sets up the blob detector parameters by creating an object to hold the parameters.
#The variable 'params' will now hold the instance of the parameters.
params = cv2.SimpleBlobDetector_Params()

#This line enables the blob filtering by their area.
#This helps consider the minimum and maximum area as the area range specified later.
params.filterByArea = True
#This sets the minimum value for the blob to be considered.
#Only the objects above the value of 50 will be processed.
params.minArea = 50
#This sets the maximum value for the blob to be considered.
#Only the objects below the value of 10000 will be processed
params.maxArea = 10000

#This line enables the blob filtering based on color.
#Blobs that are represented by a certain pixel that is specified below will be the only ones considered.
params.filterByColor = True
#This sets a color that the detector will look for in a binary/grayscale image, which is 255 (white).
#Only white pixels will be considered as blobs.
params.blobColor = 255

#This enables the filter based on the circularity of a blob.
#Blobs that have a reasonable degree of circularity will be considered.
params.filterByCircularity = True
#This sets the minimum circularity value that a blob must be to be detected.
#In this case, only blobs with 0.1 circularity and above will be considered.
params.minCircularity = 0.1

#This enables filtering on how convex a blob is.
#Blobs that do not have a significant inward curves or deep indentations will be considered.
params.filterByConvexity = True
#This sets the minimum convexity that a blob must have.
#Only blobs that have 0.1 and above convex value will be considered as a blob.
params.minConvexity = 0.1

#The function below creates the instance of the SimpleBlobDetector using the params variable.
#All the filters that have been set earlier are applied and is being held by 'detector'.
detector = cv2.SimpleBlobDetector_create(params)

#Using the 'combined_mask' binary image, it will detect blobs using the variable instance of the blob detector, 'detector.'
#The detected blobs that align with the filters will be stored in the 'keypoints' list.
keypoints = detector.detect(combined_mask)

#This helps copy the original RGB image, then stored in 'blobs'
#This will be useful for printing the blobs along with the original image.
blobs = rgb_img.copy()
#This function helps draw the blobs in red circles, with the stored information of its areas in the keypoints list.
#The flag 'DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS' will draw circles on the keypoints with a size proportional to the property of keypoints.
blob_circles = cv2.drawKeypoints(blobs, keypoints, None, (255, 0, 0), cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)


#This function specifies the size of the image to be displayed.
#This creates a figure and uses figsize to specify the width and height in inches.
plt.figure(figsize=(12, 6))
#Since pyplot mainly works with plot generation, normally axes will be seen.
#This function turns of the display of the axes, mainly to focus on the image displayed.
plt.axis('off')
#This function specifies which stored image will be displayed, in this case, it is the Blob Detection Image.
#The variables inputted will be rendered as a pseudocolor image (Matplotlib, 2025b).
plt.imshow(blob_circles)
#This function adds a title to the displayed image.
#In this title, it adds a description and context to its viewer.
plt.title('Simple Blob Detection')
#This is the last function to be executed after all the plotting commands.
#This basically shows the plotted image and the result of all our plotting commands done earlier.
plt.show()
```

```python
#This line of code just provides details for the output below.
#This will be a looping output, hence the need for a title and context for its outputs.
print("Detected Blob Details:")

#This is a loop that displays the properties of the blobs drawn.
#The keypoints list is utilized here and accessed for each iteration to print its details.
for i, kp in enumerate(keypoints):
    #This gets the coordinates of the blob center as a float type.
    #Each item in the list will have its stored location for printing later.
    x, y = kp.pt             # Coordinates of the blob center (as float)
    #This gets the size or diameter of the blob in keypoints.
    #Each item in the list will have its stored size for printing later.
    size = kp.size
    #This just basically prints the number of the blob, its location thorough coordinates, and its size.
    #Basically, it prints what have been gathered above.
    print(f"Blob {i + 1}: Location = ({x:.2f}, {y:.2f}), Size = {size:.2f}")
```

COMBINING TECHNIQUES: HIGHLIGHTING OBJECTS WITH CONTOURS

```python
#This line of code sets the kernel for the final image that utilizes multiple techniques in performing image processing.
#In here, a 2x2 kernel is used for a finer operation of eroding and dilating.
final_kernel = np.ones((2,2), np.uint8)

#This utilizes the bitwise OR operation learned in Color Filtering under Masking.
#This combines the otsu_thresh image and the combined_mask, which is a mix of Automatic Thresholding and Color Filtering under Masking.
#The final mask intends to capture areas in the Otsu's Method Thresholding that was not captured in the Combined Mask of Color Filtering.
final_mask = cv2.bitwise_or(otsu_thresh, combined_mask)

#This function specifies the size of the image to be displayed.
#This creates a figure and uses figsize to specify the width and height in inches.
plt.figure(figsize=(12, 6))
#Since pyplot mainly works with plot generation, normally axes will be seen.
#This function turns of the display of the axes, mainly to focus on the image displayed.
plt.axis('off')
#This function specifies which stored image will be displayed, in this case, it is the Final Mask Image.
#Cmap is still present in the 'imshow' parameter as this is a grayscale image, without this the image would appear yellowish rather than grayscale.
plt.imshow(final_mask, cmap='gray')
#This function adds a title to the displayed image.
#In this title, it adds a description and context to its viewer.
plt.title('Final Mask')
#This is the last function to be executed after all the plotting commands.
#This basically shows the plotted image and the result of all our plotting commands done earlier.

plt.show()
```

```python
#This line intends to dilate the mask and widen its white areas.
#This uses 2 iterations using the final kernel (a 2x2 kernel) for a finer dilation and possibly refined white spaces.
mask_dilate = cv2.dilate(final_mask, final_kernel, iterations=2)

#This line of code performs an AND operation to the mask_dilate and the adapt_thresh.
#This joins the common feature of the Adaptive Threshold and the dilated mask above, storing it in the combined variable.
combined = cv2.bitwise_and(mask_dilate, adapt_thresh)

#This line intends to erode the combined mask and edges to broaden darker areas.
#This uses 1 iterations using the combined variable and the final_kernel (a 2x2 kernel) for a finer erosion and possibly refined thinned edges.
combined_erode = cv2.erode(combined, final_kernel, iterations=1)

#This performs an AND operation to an inversed 'combined_erode' and inversed 'adapt_thresh'.
#This captures the important areas of the two images, which preserves the edges of the image to reveal shapes and features later.
for_final = cv2.bitwise_and(cv2.bitwise_not(combined_erode), cv2.bitwise_not(adapt_thresh))

#This function specifies the size of the image to be displayed.
#This creates a figure and uses figsize to specify the width and height in inches.
plt.figure(figsize=(12, 6))
#Since pyplot mainly works with plot generation, normally axes will be seen.
#This function turns of the display of the axes, mainly to focus on the image displayed.
plt.axis('off')
#This function specifies which stored image will be displayed, in this case, it is the Combined Techniques Image.
#Cmap is still present in the 'imshow' parameter as this is a grayscale image, without this the image would appear yellowish rather than grayscale.
plt.imshow(for_final, cmap='gray')
#This function adds a title to the displayed image.
#In this title, it adds a description and context to its viewer.
plt.title('Combined Otsu Method, Color Filtering (Mask and Operations), and Adaptive Threshold')
#This is the last function to be executed after all the plotting commands.
#This basically shows the plotted image and the result of all our plotting commands done earlier.
plt.show()
```

```python
#This line finds the contours of the white shapes in the 'for_final' image.
#The 'RETR_EXTERNAL' function gets the outermost contours while ignoring the inner holes.
#The 'CHAIN_APPROX_SIMPLE' function reduces the points in the contour.
img_contours, _ = cv2.findContours(for_final, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

#This copies the original image 'rgb_img' to 'final_image'.
#This will be used later to display the image along with the rectangles to emphasize recgonized features.
final_image = rgb_img.copy()

#This sets the minimum value for the contour.
#Only the objects above the value of 100 will be processed.
min_area = 100
#This sets the maximum value for the contour.
#Only the objects below the value of 10000 will be processed.
max_area = 10000

#This creates an empty list/array for storing object measurements or shape properties.
#This will also be utilized for printing the stored information for later.
object_features = []

#This sets a loop in 'img_contours'.
#Basically, this goes through each contour (contour) found in the image.
for contour in img_contours:
    #This calculates the area of the contour.
    #Basically, it gets the number of pixels contained in the contour then stores it in the variable 'area'.
    area = cv2.contourArea(contour)

    #Like any other if statements, this only continues processing if the area is within the area range defined earlier.
    #Simply, the contour will only be recognized when it is between the minimum and maximum area.
    if min_area <= area <= max_area:

        #This line gets the smallest rectangle that can surround the current contour.
        #The variables on the left side gets the x and y values of the top-left corner of the rectangle, then it height and width.
        x, y, w, h = cv2.boundingRect(contour)
        #The second parameter of the image provides the coordinates for the top-left corner of the rectangle, while the thirs parameter provides the bottom-right corner of the rectangle.
        #This adds a green rectangle on the final image, with a 2 pixel wide line.
        cv2.rectangle(final_image, (x, y), (x + w, y + h), (0, 255, 0), 2)

        #Similar to the code above, this gets the smallest rectangle surrounding the contour.
        #But on the other hand, using 'minAreaRect' allows for rotation to fit the contour perfectly.
        #Rect will store a list of the information for the rotated bounding rectangle.
        rect = cv2.minAreaRect(contour)
        #The stored 'rect' information will be utilized here to become vertices of a rotated rectangle.
        #Box becomes a NumPy array to represent the vertices and coordinates of the rotated rectangle.
        box = cv2.boxPoints(rect)
        #As the data type for the coordinates was stored as a float, this converts its data type to an integer pixel.
        #This is needed as the function to draw the rectangle later expects this data type.
        box = box.astype(np.intp)
        #This utilizes the vertices and coordinates of box.
        #This adds a blue rectangle around the shape, with still a 2 pixel wide line.
        cv2.drawContours(final_image, [box], 0, (255, 0, 0), 2)

        #The function 'moments' helps calculate features such as the center of mass of the object and its area (OpenCV, 2025).
        #Basically, this line computes a dictionary of raw moments for the contour that will be necessary for determining the centroid for later.
        M = cv2.moments(contour)
        #The "m00" is the area of the contour, and checks if the area is not equal to zero.
        #This helps avoid the zero division error, and executes the else statement if the contour area is zero.
        if M["m00"] != 0:
            #The line of code below basically calculates for the x-coordinate of the centroid.
            #Because the resulting number data type will be a float, it is converted to an integer.
            cx = int(M["m10"] / M["m00"])
            #The line of code below basically calculates for the y-coordinate of the centroid.
            #Because the resulting number data type will be a float, it is converted to an integer.
            cy = int(M["m01"] / M["m00"])
            #This draws a circle on the final image that utilizes the computed x and y coordinates 'cx' and 'cy' as the center coordinates with a radius of 4 pixels.
            #This blue circle will be filled instead of outlined because of the value '-1' on its thickness parameter.
            cv2.circle(final_image, (cx, cy), 4, (0, 0, 255), -1)

        #This is the else statement that will be executed once the contour area is 0.
        #The x and y coordinates of the contour is 0 to avoid empty contours.
        else: cx, cy = 0, 0


        #This find the total length around the contour, or its perimeter.
        #The function uses the contour as its inputs and a flag that indicates whether the curve is closed or not (OpenCV, 2025).
        perimeter = cv2.arcLength(contour, True)
        #The width of the rectangle is divided by its height, then stored in the 'aspect_ratio' variable.
        #This gives the ratio of the contour, whether its shape is tall or wide.
        aspect_ratio = float(w) / h
        #Area is divided by the width and height multiplied together.
        #This tells how much the bounding rectangle is filled by the contour, which value present is between 0 and 1.
        extent = area / (w * h)
        #The 'convexHull' function calculates the convex hull of the given contour.
        #Simply, this line finds the tighest convex to bound the contour.
        hull = cv2.convexHull(contour)
        #Using the calculated hull from the line above, the function 'contourArea' will calculate the area of the polygon.
        #Basicallym this line measure the total area occupied by the convex hull.
        hull_area = cv2.contourArea(hull)
        #This divides the area, that has been converted into float, to be divided to the computed 'hull_area'.
        #The if statement is to avoid a zero division error.
        #This simply computes for the solidity of the object, describing how filled in it is.
        solidity = float(area) / hull_area if hull_area != 0 else 0
        #Using the area of the contour multiplied by a constant, this is divided by the NumPy pi, and squarerooted to get the diameter of the contour.
        #The calculation is then stored in 'equi_diameter' for printing later.
        equi_diameter = np.sqrt(4 * area / np.pi)
        #Utilizing the stored information on the rotated bounding rectangle, it accesses the rotation angle of the rectangle.
        #This will then be stored for printing later to give context on the object's orientation.
        angle = rect[2]

        #This line of code appends the information stored per object in the object_features array list.
        #This takes the current information on the contour to be stored and printed for later for extra information.
        object_features.append({
            "Area": area,
            "Perimeter": perimeter,
            "Aspect Ratio": aspect_ratio,
            "Extent": extent,
            "Solidity": solidity,
            "Equivalent Diameter": equi_diameter,
            "Orientation Angle": angle,
            "Centroid": (cx, cy)
        })        
```
